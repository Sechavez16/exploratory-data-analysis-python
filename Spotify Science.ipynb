# Criterio de eliminación de columnas de texto

### ¿Por qué eliminamos columnas con muchos valores únicos y texto libre?

**Aclaración conceptual importante:**

Las columnas de texto **SÍ son categóricas** — sus valores son etiquetas, no números. 
Lo que las hace problemáticas es su **alta cardinalidad**, no el hecho de ser categóricas.

###  Tres conceptos clave

1. **Categórica:** variable cuyos valores son etiquetas (no números).
2. **Cardinalidad:** cuántos valores únicos distintos tiene la columna.
3. **Alta cardinalidad:** tiene muchísimos valores únicos, casi uno por fila.

###  Por qué las de alta cardinalidad no sirven como features

- **Los modelos solo entienden números.** El texto hay que convertirlo primero.
- **One-Hot explota la dimensionalidad.** Codificar 73.000 nombres de canciones distintos crearía 73.000 columnas nuevas → revienta memoria y sobreajuste.
- **No hay patrón repetible.** Si cada valor aparece 1 sola vez, el modelo no puede generalizar.
- **Aunque son categóricas, se comportan como IDs.** No aportan información predictiva.

### Criterio aplicado al dataset

| Columna | Únicos | ¿Eliminar? | Razón |
|---------|--------|-----------|-------|
| `Unnamed: 0` | 114.000 |  Eliminar | Índice del CSV, no aporta |
| `track_id` | 89.741 |  Eliminar | ID único, no aporta patrón |
| `track_name` | 73.608 |  Eliminar | Categórica de alta cardinalidad |
| `album_name` | 46.589 |  Eliminar | Categórica de alta cardinalidad |
| `artists` | 31.437 |  Eliminar | Categórica de alta cardinalidad |
| `track_genre` | 114 |  Conservar | Categórica de baja cardinalidad, se repite |
| `key`, `mode`, `time_signature` | 2–12 |  Conservar | Categóricas de baja cardinalidad |

###  Regla mental

> Si una columna de texto tiene **más de 50 valores únicos** y **cada valor aparece pocas veces**, 
> es una **categórica de alta cardinalidad**. No se puede codificar sin explotar la 
> dimensionalidad ni aporta patrón predictivo. Se elimina o se transforma en features 
> derivadas (longitud, palabras clave, número de colaboradores).

###  Distinción clave

- **No es que "no sean categóricas"** → sí lo son.
- **Es que son categóricas de ALTA CARDINALIDAD** → y por eso no se pueden usar directo.
- **Baja cardinalidad** (2-50) → codificar y usar.
- **Alta cardinalidad** (> 200) → eliminar o transformar.

##  Criterio de asignación de tipos: entender qué representa cada columna

La clave no es mirar el **dtype** que pandas le asigna automáticamente, sino 
**entender qué representa la columna en el mundo real** y asignarle el tipo adecuado.

###  Pregunta clave

Para cada columna, me pregunto:

> **¿Qué representa esta columna en la realidad?**

- Si representa **categorías/etiquetas** → dtype `category` u `object`.
- Si representa **magnitudes/cantidades** → dtype `int` o `float`.

###  Aplicado al dataset

| Columna | Representa | Dtype correcto |
|---------|-----------|----------------|
| `popularity` | Nivel de popularidad (0–100) | Numérica |
| `duration_ms` | Duración en milisegundos | Numérica |

| `acousticness` | Qué tan acústica es (0–1) | Numérica |

| `liveness` | Presencia de audiencia (0–1) | Numérica |

###  Ojo con las confusiones típicas

- **`time_signature` NO es duración.** Es compás musical (estructura rítmica).
- **`duration_ms` SÍ es duración.** Es tiempo real en milisegundos.
- **`key` NO es un número real.** Es una etiqueta de tonalidad disfrazada de int.
- **`mode` NO es una magnitud.** Es mayor (1) o menor (0), una etiqueta binaria.

###  Regla mental

 **No mira el dtype. Mira el significado.**
 
 - Si los valores son **etiquetas** → `category`.
 - Si los valores son **magnitudes reales** (se pueden sumar, promediar, comparar) → numérica.

# Balancear
Se hace neceasrio balancear, hay una diferencia avismal entre es hit o no es hit
hay mas en no es hit que en es hit, y eso significa que se va peder la clse minoritaria

````python
import numpy as np
from sklearn.utils.class_weight import compute_class_weight

pesos = compute_class_weight(
    class_weight='balanced',
    classes=np.array([0, 1]),      # ← array de numpy
    y=y_train
)

class_weight_dict = {0: pesos[0], 1: pesos[1]}
print("Class weights:", class_weight_dict)
````

##  Manejo del desbalanceo de clases

###  Diagnóstico del desbalanceo

Después de dividir el dataset en train/test con `stratify=y`, se verificó 
la proporción de cada clase en el target (`hit`):

| Clase | Significado | Proporción |
|-------|-------------|-----------|
| 0 | No hit (popularity ≤ 70) | 95.73% |
| 1 | Hit (popularity > 70) | 4.27% |

**Ratio entre clases:** ~22:1 (por cada hit hay ~22 no-hits).

**Nivel de desbalanceo:**  Fuerte (entre moderado y fuerte).

###  Por qué es un problema

Si el modelo predice **siempre "no hit"** (clase 0), acierta el **95.7%** 
de las veces sin haber aprendido nada útil. Esto se llama **"accuracy tramposa"**:

- Un modelo "tonto" que siempre dice 0 → accuracy = 95.7%.
- Pero su **recall** en la clase hit sería 0.
- Su **f1-score** sería 0.

Por eso **nunca se debe usar solo `accuracy`** en problemas desbalanceados.

###  Estrategias de balanceo

Existen tres caminos principales:

| Técnica | Qué hace | Ventaja | Desventaja |
|---------|----------|---------|------------|
| **`class_weight`** | No modifica datos. Ajusta la pérdida durante el entrenamiento. | Simple, no inventa datos | Menos agresivo |
| **SMOTE** | Crea ejemplos sintéticos de la clase minoritaria. | Datos balanceados | Puede inventar combinaciones irreales |
| **Undersampling** | Elimina ejemplos de la clase mayoritaria. | Rápido | Pierde información |

###  Técnica elegida: `class_weight`

Se eligió **`class_weight`** por las siguientes razones:

1. **No modifica los datos.** No hay riesgo de inventar combinaciones irreales.
2. **Simple de implementar** en Keras (`class_weight=class_weight_dict`).
3. **Funciona bien con `binary_crossentropy`.**
4. **Con desbalanceo 22:1, suele ser suficiente.**

````python
import tensorflow as tf
from tensorflow import keras
````

````python
# Verificar tipo
print("Tipo X_train:", type(X_train))
print("Tipo y_train:", type(y_train))

# Convertir si son DataFrames/Series
if hasattr(X_train, 'values'):
    X_train = X_train.values.astype('float32')
    X_test = X_test.values.astype('float32')

y_train = y_train.values.astype('float32')
y_test = y_test.values.astype('float32')

print("\nDespués de convertir:")
print("X_train dtype:", X_train.dtype, "shape:", X_train.shape)
print("y_train dtype:", y_train.dtype, "shape:", y_train.shape)
````

````python
model = keras.Sequential([
    keras.Input(shape=(X_train.shape[1],)), #Fatures o columnas de X_Train son 140 con el codificador
    keras.layers.Dense(64, activation='relu'),#Capa 2 de 64 neuronas ocultas
    keras.layers.Dense(32, activation='relu'),#Capa 3 de 32 neuronas ocultas
    keras.layers.Dense(1, activation='sigmoid')#Capa 1 de salida
])

model.summary()#mostramos la aruitectura completa
````

````python
#Compilamso el modelo
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy', keras.metrics.Precision(name='precision'),
             keras.metrics.Recall(name='recall'), keras.metrics.AUC(name='auc')]
)
````

````python
history = model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=20,
    batch_size=32,
    class_weight=class_weight_dict    #Escogemos lo del balanceo
)
````
